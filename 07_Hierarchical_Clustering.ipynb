# Hierarchical Clustering

Customer segmentation using Agglomerative Hierarchical Clustering (based on the linked GitHub notebook).

## 1. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the dataset

This uses the `KMeans.csv` dataset from the same GitHub repository. If you are running locally, keep the CSV in the same folder as this notebook.

```python
dataset_url = 'https://raw.githubusercontent.com/nishithkotak/machine-learning/master/KMeans.csv'
dataset = pd.read_csv(dataset_url)
dataset.head()
```

## 3. Select features

The original notebook uses columns 1 and 2: Annual Income and Spending Score.

```python
X = dataset.iloc[:, [1, 2]].values
print(X[:5])
```

## 4. Visualize the dendrogram

Ward linkage minimizes the variance within each cluster. The dendrogram helps choose the number of clusters.

```python
import scipy.cluster.hierarchy as sch

dendrogram = sch.dendrogram(sch.linkage(X, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```

## 5. Apply Agglomerative Hierarchical Clustering

The original exercise uses 5 clusters and Ward linkage.

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=5, metric='euclidean', linkage='ward')
y_hc = hc.fit_predict(X)
print(y_hc)
```

## 6. Visualize the clusters

```python
plt.figure(figsize=(8, 6))
for cluster in np.unique(y_hc):
    plt.scatter(X[y_hc == cluster, 0], X[y_hc == cluster, 1], label=f'Cluster {cluster + 1}')

plt.title('Clusters of customers (Hierarchical Clustering)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```
