```python
#Load dataset
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Advertising Budget and Sales.csv")

print("Original Data:")
print(df.head())

print("Shape of dataset:")
print(df.shape)
```

```python
#clean the data
df = df.drop(columns=["Unnamed: 0"])


print("\nMissing values:")
print(df.isnull().sum())


df = df.dropna()

print("\nShape after cleaning:")
print(df.shape)
```

```python
#select predictors and target
X = df[[
    "TV Ad Budget ($)",
    "Radio Ad Budget ($)",
    "Newspaper Ad Budget ($)"
]].values


y = df["Sales ($)"].values

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
#split the data
np.random.seed(47)

indices = np.random.permutation(len(X))

split = int(0.8 * len(X))

X_train = X[indices[:split]]
y_train = y[indices[:split]]

X_test = X[indices[split:]]
y_test = y[indices[split:]]

print("Training samples", len(X_train))
print("Testing samples", len(X_test))
```

```python
#adding intecept column
X_train_with_intercept = np.column_stack(
    (np.ones(len(X_train)), X_train)
)

X_test_with_intercept = np.column_stack(
    (np.ones(len(X_test)), X_test)
)
```

```python
#calculating regression coeffients
X_transpose = X_train_with_intercept.T
XTX = X_transpose.dot(X_train_with_intercept)
XTX_inverse = np.linalg.inv(XTX)
XTy = X_transpose.dot(y_train)
beta = np.dot(XTX_inverse,XTy)
```

```python
print("Regression Coefficients:")
print("Intercept :", beta[0])
print("TV        :", beta[1])
print("Radio     :", beta[2])
print("Newspaper :", beta[3])
```

```python
#predictions
y_pred = np.dot(X_test_with_intercept, beta)

print("First 10 predictions:")

for i in range(10):
  print(
      "Actual:", y_test[i],
      "Predicated:", y_pred[i]
  )
```

```python
#calculate time
errors = y_test - y_pred

MAE = np.mean(np.abs(errors))

print("MAE:", MAE)
```

```python
#calulate MSE
MSE = np.mean(errors**2)

print("MSE:", MSE)
```

```python
#calulate RMSE
RMSE = np.sqrt(MSE)

print("RMSE:", RMSE)
```

```python
#calulate R-square
y_mean = np.mean(y_test)

SS_res = np.sum((y_test - y_pred)**2)
SS_tot = np.sum((y_test - y_mean)**2)

R_squared = 1 - (SS_res / SS_tot)

print("R-squared:", R_squared)
```

```python
#finding predictor with largest co-efficent
predictor_names = [
    "TV",
    "Radio",
    "Newspaper"
]

predictor_coefficients = beta[1:]

largest_index = np.argmax(predictor_coefficients)

print("Predictor with largest coefficient:")
print(predictor_names[largest_index])

print(
    "Largest coefficient:",
    predictor_coefficients[largest_index]
)
```

```python
#final regression equation
print("Regression Equation:")

print(
    "Sales =",
    beta[0],
    "+", beta[1], "* TV",
    "+", beta[2], "* Radio",
    "+", beta[3], "* Newspaper"
)
```
